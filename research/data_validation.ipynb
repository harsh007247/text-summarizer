```python
from pathlib import Path
from dataclasses import dataclass

@dataclass(frozen=True)
class datavalidationConfig:
    root_dir: Path 
    STATUS_FILE: str
    ALL_REQUIRED_FILES: list
```

```python
from textsummarizer.constants import *
from textsummarizer.utils.common import read_yaml,creat_directories
```

```python
class ConfigurationManager:
    def __init__(self,
                config_filepath=CONFIG_FILE_PATH,
                params_filepath=PARAMS_FILE_PATH):

        self.config=read_yaml(config_filepath)
        self.params=read_yaml(params_filepath)
        creat_directories([self.config.artifacts_root])

    def get_data_validation_config(self) -> datavalidationConfig:
        config=self.config.data_validation
        creat_directories([config.root_dir])

        data_validation_config= datavalidationConfig(
            root_dir=config.root_dir,
            STATUS_FILE=config.STATUS_FILE,
            ALL_REQUIRED_FILES= config.ALL_REQUIRED_FILES,
            )
        return data_validation_config
```

```python
import os
from textsummarizer.logging import logger
```

```python
class DataValidation:
    def __init__(self,config: datavalidationConfig):
        self.config=config

    def validate_all_files_exists(self)->bool:
        try:
            validate_status=None
            all_files=os.listdir(os.path.join("artifacts","data_ingestion","samsum_dataset"))
            for file in all_files:
                if file not in self.config.ALL_REQUIRED_FILES:
                    validate_status=False
                    with open(self.config.STATUS_FILE,"w") as f:
                        f.write(f"validation status: {validate_status}")

                else:
                    validate_status=True
                    with open(self.config.STATUS_FILE,"w") as f:
                        f.write(f"validation status: {validate_status}")
            return validate_status

        except Exception as e:
            raise e

    
```

```python
try:
    config= ConfigurationManager()
    data_validation_config=config.get_data_validation_config()
    data_validation=DataValidation(config=data_validation_config)
    data_validation.validate_all_files_exists()


except Exception as e:
    raise e
```
